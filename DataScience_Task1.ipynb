```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
#Name : Srushti Ambre
# Task 1: Titanic EDA
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

DATA_PATH = "titanic.xlsx"

def load_and_clean_data(path: str) -> pd.DataFrame:
    """Load Titanic data and handle missing values."""
    df = pd.read_excel(path)
    # Filling missing values
    df["Age"] = df["Age"].fillna(df["Age"].mean())
    df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
    return df

def plot_survival_by_gender(df: pd.DataFrame) -> None:
    plt.figure(figsize=(7, 5))
    sns.countplot(x="Survived", hue="Sex", data=df)
    plt.title("Survival Count by Gender")
    plt.xlabel("Survived (0 = No, 1 = Yes)")
    plt.ylabel("Passenger count")
    plt.show()

def plot_survival_rate_by_class(df: pd.DataFrame) -> None:
    plt.figure(figsize=(7, 5))
    sns.barplot(x="Pclass", y="Survived", data=df)
    plt.title("Survival Rate by Passenger Class")
    plt.xlabel("Passenger class (Pclass)")
    plt.ylabel("Survival rate")
    plt.show()

def plot_correlation_heatmap(df: pd.DataFrame) -> None:
    plt.figure(figsize=(10, 8))
    numeric_df = df.select_dtypes(include=['number'])
    sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
    plt.title("Correlation Heatmap of Titanic Features")
    plt.show()

def plot_age_distribution(df: pd.DataFrame) -> None:
    plt.figure(figsize=(8, 5))
    sns.histplot(df['Age'], kde=True, color='teal')
    plt.title("Distribution of Passenger Ages")
    plt.show()

def main() -> None:
    """Run all 4 Titanic EDA plots."""
    sns.set_theme(style="whitegrid")
    df = load_and_clean_data(DATA_PATH)

    print("Generating Survival by Gender...")
    plot_survival_by_gender(df)

    print("Generating Survival Rate by Class...")
    plot_survival_rate_by_class(df)

    print("Generating Correlation Heatmap...")
    plot_correlation_heatmap(df)

    print("Generating Age Distribution...")
    plot_age_distribution(df)

if __name__ == "__main__":
    main()
```

Executive Summary of Titanic Data Analysis
Gender Bias: Females had a significantly higher survival count than males, following the "women and children first" protocol.

Class Advantage: There is a direct correlation between social class and survival. 1st Class passengers had the highest survival rate, while 3rd Class had the lowest.

Age Spread: The majority of passengers were between 20 and 40 years old, with a notable number of infants who also saw higher survival priority.

Feature Relations: The heatmap shows that Fare and Pclass are strongly related—higher fares (1st class) correlate with higher survival chances.
